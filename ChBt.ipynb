```python
from langchain_groq import ChatGroq
llm = ChatGroq(api_key=api_key, model_name=LLAMA_MODEL)

def simplechat(userprompt):
  response= llm.invoke(userprompt)
  print(response)
  return response.content
```

```python
simplechat("How to use langchain groq for RAG model")
```

```python
from langchain_core.messages import HumanMessage, AIMessage

converse= []
def simplechatwithmem(userprompt):
  converse.append(HumanMessage(content=userprompt))
  response=llm.invoke(converse)
  converse.append(AIMessage(content=response.content))
  return response.content
```

```python
mes = simplechatwithmem("How to use langchain groq for RAG model")
print(mes)
```

```python
resp= simplechatwithmem("Give me sample real world usage and implementation plan of the same")
print(resp)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt= ChatPromptTemplate (messages = [
        ("system","You are a helpful assitant"),
        ("human", "{userprompt}")
    ]
)

chain = prompt | llm
chain.invoke({"userprompt":"Share me the efficient ways to implement RAG model"})
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt= ChatPromptTemplate (messages = [
        ("system","You are a helpful assitant"),
        ("placeholder", "{chat_history}"),
        ("human", "{userprompt}")
    ]
)

chainwithmem = prompt | llm
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory

store={}

def get_chat_history(sessionid:str):
  if sessionid not in store:
    store[sessionid]=ChatMessageHistory()
  return store[sessionid]

```

```python
from langchain_core.runnables import RunnableWithMessageHistory

chatwithmemhist= RunnableWithMessageHistory(
    runnable = chainwithmem,
    get_session_history= get_chat_history,
    input_variables="userprompt",
    history_messages_key="chat_history"
)

resp= chatwithmemhist.invoke({"userprompt":"Share me the efficient ways to implement RAG model"}, {"configurable":{"session_id":"user1"}})
print(resp)



```
